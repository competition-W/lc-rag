```python
def show_the_nodes(nodes):
    for i in range(0,5):
        # 验证
        print(f"*********************THIS IS {i}*********************")
        print(nodes[i].text[:500])
```

```python
from llama_index.readers.file import FlatReader
from pathlib import Path
from llama_index.core.schema import Document
from typing import List

def get_md_file(md_file_path:str) -> List[Document]:
    '''直接加载markdown文件
    拓展：
        加载文件列表，批处理
        lazy_load()支持
    '''
    ori_markdown = FlatReader().load_data(Path(md_file_path))
    return ori_markdown
```

```python
markdown_docs = get_md_file("/mnt/chatchat/aimtoMD/AILabshiyongguide_V2.md")
from llama_index.core.node_parser import MarkdownNodeParser

parser = MarkdownNodeParser()

nodes = parser.get_nodes_from_documents(markdown_docs)
```

```python
len(nodes)
```

```python
show_the_nodes(nodes)
```

```python
# 或者带元数据和前后关系
parser = MarkdownNodeParser.from_defaults(include_metadata=True, include_prev_next_rel=True)

# 3. 对文档进行节点解析
nodes = parser.get_nodes_from_documents(markdown_docs)
```

```python
show_the_nodes(nodes)
```

```python
from llama_index.core.node_parser import SentenceSplitter

splitter = SentenceSplitter(
    chunk_size=1024,
    chunk_overlap=20,
)
nodes = splitter.get_nodes_from_documents(markdown_docs)
```

```python
show_the_nodes(nodes)
```
